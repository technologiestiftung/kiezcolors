```python
%%bash
ogr2ogr \
  --config OGR_WFS_PAGING_ALLOWED ON \
  --config OGR_WFS_PAGE_SIZE 100000 \
  -f GeoJSON \
  -t_srs EPSG:4326 \
  -select bezeich \
  output_4326.geojson \
  "WFS:https://gdi.berlin.de/services/wfs/alkis" \
  "alkis:tatsaechlichenutzungflaechen" \
  -oo VERSION=2.0.0
```

```python
%%bash
tippecanoe --output-to-directory ../static/tiles --layer "alkis" --no-tile-compression --force -B 10 '--minimum-zoom=10' '--maximum-zoom=13' ./output_4326.geojson
```
